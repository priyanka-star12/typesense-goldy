```python

QUERY = "Harry Potter"

COLLECTION_NAME = "books"

search_parameters = {
    "q": QUERY,
    "query_by": "title,authors",
    "sort_by": "publication_year:desc",
    "per_page": 10
}

search_results = client.collections[COLLECTION_NAME].documents.search(
    search_parameters
)

print("Search query:", QUERY)
print("Found:", search_results["found"])

for hit in search_results["hits"]:
    book = hit["document"]
    print(
        f'- {book["title"]} ({book["publication_year"]}) — '
        f'{", ".join(book["authors"])}'
    )
```

```python
import os
import json
import typesense
from getpass import getpass

HOST = "asxw3mkqc18idfeop-1.a1.typesense.net"
COLLECTION = "readers_catalog"

api_key = os.getenv("qWyV5x2Djj1W543yPD8mybk0pElyGOX8")
if not api_key:
    api_key = getpass("Enter your Typesense API key: ")

client = typesense.Client({
    "nodes": [{
        "host": HOST,
        "port": "443",
        "protocol": "https"
    }],
    "api_key": api_key,
    "connection_timeout_seconds": 10
})

print("Typesense client configured.")
```

```python
with open("Readers.json", "r", encoding="utf-8") as f:
    readers = json.load(f)

print("Total readers:", len(readers))
print("First reader:", readers[0])
```

```python
schema = {
    "name": COLLECTION,
    "fields": [
        {"name": "name", "type": "string"},
        {"name": "city", "type": "string", "facet": True},
        {"name": "favorite_genre", "type": "string", "facet": True},
        {"name": "books_read", "type": "int32"},
        {"name": "membership", "type": "string", "facet": True}
    ]
}

try:
    client.collections[COLLECTION].retrieve()
    print("Readers collection already exists.")
except typesense.exceptions.ObjectNotFound:
    client.collections.create(schema)
    print("Readers collection created.")
```

```python
result = client.collections[COLLECTION].documents.import_(
    readers,
    {"action": "upsert"}
)

for row in result:
    print(row)
```

```python
query = input("Enter reader name, city, or genre: ")

results = client.collections[COLLECTION].documents.search({
    "q": query,
    "query_by": "name,city,favorite_genre",
    "per_page": 20
})

print("Total matches:", results["found"])

for hit in results["hits"]:
    r = hit["document"]
    print(
        f"Name: {r['name']} | "
        f"City: {r['city']} | "
        f"Genre: {r['favorite_genre']} | "
        f"Books read: {r['books_read']} | "
        f"Membership: {r['membership']}"
    )
```

```python
city = "Sangli"

results = client.collections[COLLECTION].documents.search({
    "q": "*",
    "query_by": "name",
    "filter_by": f"city:={city}",
    "per_page": 20
})

for hit in results["hits"]:
    print(hit["document"])
```

```python
results = client.collections[COLLECTION].documents.search({
    "q": "*",
    "query_by": "name",
    "filter_by": "books_read:>=15",
    "sort_by": "books_read:desc",
    "per_page": 20
})

for hit in results["hits"]:
    r = hit["document"]
    print(r["name"], "-", r["books_read"], "books")
```
